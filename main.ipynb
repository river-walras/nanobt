```python
from nanobt.utils import load_file
import polars as pl

_, _, df_future = load_file(
    "/share/tardis/raw/binance-futures/book_ticker/BTCUSDT/binance-futures_book_ticker_2025-11-01_BTCUSDT.csv.gz",
    to_numpy=False,
)
_, _, df_spot = load_file(
    "/share/tardis/raw/binance/book_ticker/BTCUSDT/binance_book_ticker_2025-11-01_BTCUSDT.csv.gz",
    to_numpy=False,
)
```

```python
df_future = df_future.to_pandas()
```

```python
df_spot = df_spot.to_pandas()
```

```python
df = df_future.merge(
    df_spot,
    on="exch_ts",
    suffixes=("_future", "_spot"),
    how="outer",
)
df
```

```python
df = df.ffill().bfill()
df
```

```python
df['ratio'] = (df['ask_px_future'] + df['bid_px_future'])  / (df['ask_px_spot'] + df['bid_px_spot'] ) - 1
df['ratio_med'] = df['ratio'].rolling(window=200).median()


df['ratio'].mean(), df['ratio'].std(), df['ratio'].min(), df['ratio'].max()
```
